```python
import sys
sys.path.append('/home/sid/Projects/reprojection')
import json
import numpy as np
from reproject.reproject_mesh.reprojection_utils import get_transformation_matrix_opencv
```

```python
json_file = "/home/NAS-mountpoint/kinect-omni-ego/2022-10-07/at-a02/bedroom/a03/capture2/out_capture2/rgb_preprocess/mp_keypoints_3d/1665141386007.json"

# Open json file
with open(json_file, 'r') as f:
    data = json.load(f)

print(data.keys())
```

```python
pose_array = np.array(data["pose_world_landmarks"])
pose_array.shape
```

```python
cam0_to_world_pth = "/home/NAS-mountpoint/kinect-omni-ego/2022-10-07/at-a02/bedroom/calib/k2-omni/capture2/k2_rgb_cam_to_world.pkl"
world_to_cam1_pth = "/home/NAS-mountpoint/kinect-omni-ego/2022-10-07/at-a02/bedroom/calib/k2-omni/k2_omni_world_to_cam.pkl"
transform = get_transformation_matrix_opencv(cam0_to_world_pth, world_to_cam1_pth)
```

```python
transform.shape
```

```python
# transform the points using the 4x4 transformation matrix
transform
```

```python
homo_skel = np.concatenate([pose_array, np.ones((1,33,1))], axis=-1)
homo_skel = homo_skel.transpose(0,2,1).reshape(4,-1)
transformed_skel = transform @ homo_skel
transformed_skel = transformed_skel[:3].transpose(1,0).reshape(1, 33, 3)
```
